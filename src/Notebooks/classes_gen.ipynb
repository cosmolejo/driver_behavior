```python
clases = ['blinks/blinking',
       'driver_actions/drinking', 'driver_actions/hair_and_makeup',
       'driver_actions/phonecall_left', 'driver_actions/phonecall_right',
       'driver_actions/radio', 'driver_actions/reach_backseat',
       'driver_actions/reach_side', 'driver_actions/safe_drive',
       'driver_actions/talking_to_passenger',
       'driver_actions/texting_left', 'driver_actions/texting_right', 'eyes_state/close',
       'eyes_state/closing', 'eyes_state/open', 'eyes_state/opening',
       'talking/talking', 'yawning/Yawning with hand',
       'yawning/Yawning without hand']

df = df.query(f'Actividad in {clases}')
df
```

```python
clas_dict = {'blinks/blinking': 'drowsiness',
       'driver_actions/drinking' : 'eating/drinking', 
       'driver_actions/hair_and_makeup' : 'makeup', 
       'driver_actions/phonecall_left' : 'devices', 
       'driver_actions/phonecall_right': 'devices',
       'driver_actions/radio': 'devices',
       'driver_actions/reach_backseat' : 'reaching',
       'driver_actions/reach_side': 'reaching',
       'driver_actions/safe_drive': 'safe',
       'driver_actions/talking_to_passenger': 'talking_',
       'driver_actions/texting_left': 'devices', 
       'driver_actions/texting_right': 'devices', 
       'eyes_state/close': 'drowsiness_',
       'eyes_state/closing': 'drowsiness_', 
       'eyes_state/open': 'drowsiness_', 
       'eyes_state/opening': 'drowsiness_',
       'talking/talking': 'talking', 
       'yawning/Yawning with hand': 'drowsiness',
       'yawning/Yawning without hand': 'drowsiness'}

df['class'] = df.Actividad.apply(lambda x: clas_dict[x])
df['class']
```

```python
sns.histplot(data=df, y='class')
```
